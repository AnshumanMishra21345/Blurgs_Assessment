```python
import numpy as np
import matplotlib.pyplot as plt

# sample vessel observations
# using a small example dataset for now
lat = np.array([12.95, 12.96, 12.97, 12.98])
lon = np.array([80.10, 80.12, 80.14, 80.16])

# fit a simple straight line
t = np.arange(len(lat))
lat_fit = np.polyfit(t, lat, 1)
lon_fit = np.polyfit(t, lon, 1)

# predict a few steps ahead
future_t = np.arange(len(lat), len(lat) + 7)
pred_lat = np.polyval(lat_fit, future_t)
pred_lon = np.polyval(lon_fit, future_t)

# plot the trajectory
plt.figure(figsize=(8, 5))
plt.plot(lon, lat, "o-", label="Observed")
plt.plot(pred_lon, pred_lat, "--", label="Predicted")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Vessel trajectory")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
